```python
import json
```

```python
#writing a func to load data
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("data01.json")
```

```python
#write a function to display users and their connections
def display_users(data):
    print("Users and their connections : \n")
    for user in data['users']:
        print(f"ID :{user['id']}- {user['name']} is friends with :{user['friends']} and liked pages are {user['liked_pages']}")
    print("\nPages Info")
    for page in data ['pages']:
        print(f"{page['id']} : {page['name']}")

display_users(data)  
```
